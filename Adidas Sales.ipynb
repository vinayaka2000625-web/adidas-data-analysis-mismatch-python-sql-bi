```python
# Always opening a file start with "Restart the kernal and Run all cells".
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
df["Invoice Date"] = pd.to_datetime (df["Invoice Date"].str.replace("-", "/"), format='%m/%d/%Y', errors = 'coerce') 

# Replaced "-" by "/" and also converted string to date time.
```

```python
df.info()
```

```python
numeric_cols = ['Price per Unit', 'Units Sold','Total Sales','Operating Profit'] 
for col in numeric_cols:
  df[col] = (df[col].astype(str).str.replace(r'[^0-9.]', '', regex = True).str.strip())
 

    # Removed Extra space and Special Character like ("$","-" ).
```

```python
for col in numeric_cols:
     df[col] = pd.to_numeric(df[col], errors='coerce')
   # Converted to string to numeric(float).
```

```python
df.info()
```

```python
df.isnull().sum()
# Checking for nulls
```

```python
df['Price per Unit'].isna().sum()
```

```python
median_price = df['Price per Unit'].median() 
median_price
# Calculated median
```

```python
df['Price per Unit'] = df['Price per Unit'].fillna(median_price)
# Fill or replace null into median value  
```

```python
df['Price per Unit'].isna().sum()
```

```python
df
```

```python
df["Calculated Sales"] = df["Price per Unit"] * df["Units Sold"]
df["Calculated Sales"]
# Calculated Actual Sales Value
```

```python
df["Revenue Variance"] = df["Total Sales"] - df["Calculated Sales"]
df["Revenue Variance"]
# Calculated a value(Revenue Variance) difference amount between Total Sales and Calculated Sales. OR Compared with reported value
```

```python
df[["Total Sales", "Calculated Sales", "Revenue Variance"]].head()
```

```python
df["Revenue Flag"] = df["Revenue Variance"].apply(lambda x: 'Mismatch' if x != 0 else 'Correct')
df
# Classifying Revenue Records Correct vs Mismatch.
```

```python
df[["Total Sales", "Calculated Sales", "Revenue Variance", "Revenue Flag"]].head(50)
```

```python
summary = df.groupby("Revenue Flag").size().reset_index(name="count")
plt.figure(figsize=(6, 4))
plt.bar(summary["Revenue Flag"], summary["count"], color=["Green", "Red"])
plt.xlabel("Revenue Status", fontsize=11)
plt.ylabel("Total Sales", fontsize=11)
plt.title("Revenue Validation Summary", fontsize=14, fontweight="bold")
plt.grid(axis="y", linestyle="--", alpha=0.5)
for bar in plt.gca().patches:
    value = bar.get_height()
    plt.text(
        bar.get_x() + bar.get_width() / 2, value,
        str(int(value)),
        ha="center", va="bottom")
plt.gca().spines["top"].set_visible(False)
plt.gca().spines["right"].set_visible(False)
plt.tight_layout()
plt.show()

# Showed below in graphical representation of Correct values VS Mismatch
```
